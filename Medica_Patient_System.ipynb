```python
# Prosty system obsługi pacjentów - Medica+

def zarejestruj_pacjenta(imie, nazwisko, objawy):
    pacjent = {
        "imie": imie,
        "nazwisko": nazwisko,
        "objawy": objawy,
        "status": "Oczekujący na wizytę"
    }
    print(f" Sukces: Pacjent {imie} {nazwisko} został zarejestrowany w systemie Medica+.")
    return pacjent

# Przykładowe użycie
nowy_pacjent = zarejestruj_pacjenta("Anna", "Kowalska", "Ból zęba")
```
